```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.feature_selection import SelectFromModel

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

df = pd.read_csv("paddydataset.csv")

df.columns = df.columns.str.strip()

print("Original dataset shape:", df.shape)
display(df.head())


df.info()

print("\nMissing values:")
print(df.isnull().sum().sort_values(ascending=False))

print("\nDuplicate occurrences:", df.duplicated().sum())

print("\nStatistical summary:")
display(df.describe().T)


df = df.drop_duplicates().reset_index(drop=True)

print("\nShape after duplicate removal:", df.shape)


# Target distribution
plt.figure(figsize=(8, 5))
sns.histplot(df["Paddy yield(in Kg)"], kde=True)
plt.title("Distribution of Paddy Yield")
plt.xlabel("Paddy Yield (Kg)")
plt.show()


target = "Paddy yield(in Kg)"

X = df.drop(columns=[target])
y = df[target]

# Remove rows with missing target values
valid_rows = y.notna()
X = X.loc[valid_rows].copy()
y = y.loc[valid_rows].copy()

print("\nInput features:", X.shape[1])
print("Target:", target)


#  IDENTIFY NUMERICAL AND CATEGORICAL FEATURES

cat_cols = X.select_dtypes(
    include=["object", "category", "str"]
).columns.tolist()

num_cols = X.select_dtypes(
    include=np.number
).columns.tolist()

print("\nCategorical features:", cat_cols)
print("Numerical features:", len(num_cols))

#  PREPROCESSING

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("encoder", OneHotEncoder(handle_unknown="ignore"))
            ]),
            cat_cols
        ),
        (
            "numerical",
            SimpleImputer(strategy="median"),
            num_cols
        )
    ]
)


# TRAIN-TEST SPLIT

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

#  PREPROCESS TRAINING AND TESTING DATA

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("\nFeatures after encoding:", X_train_processed.shape[1])


#  FEATURE SELECTION USING RANDOM FOREST

feature_selector = SelectFromModel(
    estimator=RandomForestRegressor(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ),
    threshold=-np.inf,
    max_features=min(20, X_train_processed.shape[1])
)

feature_selector.fit(X_train_processed, y_train)

X_train_selected = feature_selector.transform(X_train_processed)
X_test_selected = feature_selector.transform(X_test_processed)

print("\nFeatures before selection:", X_train_processed.shape[1])
print("Features after selection:", X_train_selected.shape[1])


# Display selected feature names
all_feature_names = preprocessor.get_feature_names_out()

selected_mask = feature_selector.get_support()
selected_features = all_feature_names[selected_mask]

print("\nSelected features:")
print(selected_features.tolist())


model = RandomForestRegressor(
    n_estimators=200,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train_selected, y_train)

print("\nModel training completed.")


y_pred = model.predict(X_test_selected)

results = pd.DataFrame({
    "Actual Yield (Kg)": y_test.values,
    "Predicted Yield (Kg)": y_pred
})

display(results.head(10))

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("\nRandom Forest with Feature Selection")
print("MAE:", round(mae, 2), "Kg")
print("RMSE:", round(rmse, 2), "Kg")
print("R2 Score:", round(r2, 4))


plt.figure(figsize=(7, 5))

plt.scatter(y_test, y_pred, alpha=0.7)

low = min(y_test.min(), y_pred.min())
high = max(y_test.max(), y_pred.max())

plt.plot([low, high], [low, high], "r--")

plt.xlabel("Actual Yield (Kg)")
plt.ylabel("Predicted Yield (Kg)")
plt.title("Actual vs Predicted Paddy Yield")
plt.show()


importance_df = pd.DataFrame({
    "Feature": selected_features,
    "Importance": model.feature_importances_
}).sort_values(
    by="Importance",
    ascending=False
)

display(importance_df)

top_features = importance_df.head(15).sort_values("Importance")

plt.figure(figsize=(10, 7))

plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)

plt.xlabel("Feature Importance")
plt.title("Selected Feature Importance")
plt.show()

param_grid = {
    "n_estimators": [100, 200],
    "max_depth": [None, 10, 20],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2]
}

grid = GridSearchCV(
    estimator=RandomForestRegressor(
        random_state=42,
        n_jobs=1
    ),
    param_grid=param_grid,
    cv=5,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1,
    verbose=1
)

grid.fit(X_train_selected, y_train)

print("\nBest parameters:", grid.best_params_)
print("Best CV RMSE:", round(-grid.best_score_, 2), "Kg")


best_model = grid.best_estimator_

y_pred_tuned = best_model.predict(X_test_selected)

print("\nTuned Random Forest Results")
print(
    "MAE:",
    round(mean_absolute_error(y_test, y_pred_tuned), 2),
    "Kg"
)

print(
    "RMSE:",
    round(np.sqrt(mean_squared_error(y_test, y_pred_tuned)), 2),
    "Kg"
)

print(
    "R2 Score:",
    round(r2_score(y_test, y_pred_tuned), 4)
)


comparison = pd.DataFrame({
    "Metric": ["MAE", "RMSE", "R2 Score"],
    "Random Forest": [
        mean_absolute_error(y_test, y_pred),
        np.sqrt(mean_squared_error(y_test, y_pred)),
        r2_score(y_test, y_pred)
    ],
    "Tuned Random Forest": [
        mean_absolute_error(y_test, y_pred_tuned),
        np.sqrt(mean_squared_error(y_test, y_pred_tuned)),
        r2_score(y_test, y_pred_tuned)
    ]
})

display(comparison.round(4))
```
